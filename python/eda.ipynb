```python
#Step 8: Convert Dates
customers["JoinDate"] = pd.to_datetime(customers["JoinDate"])

sales["OrderDate"] = pd.to_datetime(sales["OrderDate"])

inventory["LastRestocked"] = pd.to_datetime(
    inventory["LastRestocked"]
)
```

```python
#Step 9: Merge Tables
sales_data = sales.merge(
    customers,
    on="CustomerID",
    how="left"
)

sales_data = sales_data.merge(
    products,
    on="ProductID",
    how="left"
)
```

```python
sales_data.head()
```

```python
#Step 10: Create Business KPIs
#Total Revenue :
sales_data["Sales"].sum()
```

```python
#Total Profit :
sales_data["Profit"].sum()
```

```python
#Total Customers :
customers["CustomerID"].nunique()
```

```python
#Total Products :
products["ProductID"].nunique()
```

```python
#Average Order Value :
sales_data["Sales"].mean()
```

```python
#Step 11: Top 10 Products
top_products = (
    sales_data
    .groupby("ProductName")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
#Step 12: Sales by Category
sales_category = (
    sales_data
    .groupby("Category")["Sales"]
    .sum()
)

sales_category
```

```python
#Step 13: Monthly Revenue
sales_data["Month"] = (
    sales_data["OrderDate"]
    .dt.to_period("M")
)

monthly_sales = (
    sales_data
    .groupby("Month")["Sales"]
    .sum()
)

monthly_sales
```

```python
#Step 14: Create Charts
#Monthly Sales
monthly_sales.plot()

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
#Category Sales
sales_category.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()
```

```python
#Top Products
top_products.plot(kind="bar")

plt.title("Top Selling Products")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```
