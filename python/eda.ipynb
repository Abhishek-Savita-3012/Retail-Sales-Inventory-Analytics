```python
#Step 8: Convert Dates
customers["JoinDate"] = pd.to_datetime(customers["JoinDate"])

sales["OrderDate"] = pd.to_datetime(sales["OrderDate"])

inventory["LastRestocked"] = pd.to_datetime(
    inventory["LastRestocked"]
)
```

```python
#Step 9: Merge Tables
sales_data = sales.merge(
    customers,
    on="CustomerID",
    how="left"
)

sales_data = sales_data.merge(
    products,
    on="ProductID",
    how="left"
)
```

```python
sales_data.head()
```

```python
#Step 10: Create Business KPIs
#Total Revenue :
sales_data["Sales"].sum()
```

```python
#Total Profit :
sales_data["Profit"].sum()
```

```python
#Total Customers :
customers["CustomerID"].nunique()
```

```python
#Total Products :
products["ProductID"].nunique()
```

```python
#Average Order Value :
sales_data["Sales"].mean()
```

```python
#Step 11: Top 10 Products
top_products = (
    sales_data
    .groupby("ProductName")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
#Step 12: Sales by Category
sales_category = (
    sales_data
    .groupby("Category")["Sales"]
    .sum()
)

sales_category
```

```python
#Step 13: Monthly Revenue
sales_data["Month"] = (
    sales_data["OrderDate"]
    .dt.to_period("M")
)

monthly_sales = (
    sales_data
    .groupby("Month")["Sales"]
    .sum()
)

monthly_sales
```

```python
#Step 14: Create Charts
#Monthly Sales
monthly_sales.plot()

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
#Category Sales
sales_category.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()
```

```python
#Top Products
top_products.plot(kind="bar")

plt.title("Top Selling Products")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```

```python
#Step 1: Customer Purchase Distribution (How much do customers usually spend?)
customer_sales = (
    sales_data
    .groupby("CustomerID")["Sales"]
    .sum()
)

customer_sales.describe()
```

```python
customer_sales.plot(
    kind="hist",
    bins=30
)

plt.title("Customer Purchase Distribution")
plt.xlabel("Total Spending")
plt.ylabel("Number of Customers")

plt.show()
```

```python
#Step 2: Top 10 Customers
top_customers = (
    sales_data
    .groupby("CustomerName")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers
```

```python
top_customers.plot(kind="bar")

plt.title("Top 10 Customers")
plt.xlabel("Customer")
plt.ylabel("Revenue")

plt.show()
```

```python
#Step 3: Category-wise Profit
category_profit = (
    sales_data
    .groupby("Category")["Profit"]
    .sum()
)

category_profit
```

```python
category_profit.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
#Step 4: Product Profitability
product_profit = (
    sales_data
    .groupby("ProductName")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

product_profit
```

```python
product_profit.plot(kind="bar")

plt.title("Top Profitable Products")
plt.xlabel("Product")
plt.ylabel("Profit")

plt.show()
```

```python
#Step 5: State-wise Revenue
state_sales = (
    sales_data
    .groupby("State")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

state_sales
```

```python
#Step 6: Customer Age Distribution
customers["Age"].plot(
    kind="hist",
    bins=20
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Customers")

plt.show()
```

```python
#Step 7: Gender Distribution
gender_distribution = (
    customers["Gender"]
    .value_counts()
)

gender_distribution.plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
#Step 8: Monthly Revenue Trend
monthly_sales = (
    sales_data
    .groupby(
        sales_data["OrderDate"].dt.to_period("M")
    )["Sales"]
    .sum()
)

monthly_sales.plot()

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
#Step 9: Monthly Profit Trend
monthly_profit = (
    sales_data
    .groupby(
        sales_data["OrderDate"].dt.to_period("M")
    )["Profit"]
    .sum()
)

monthly_profit.plot()

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
#Step 10: Correlation Analysis (Does selling more units increase revenue and profit?)
correlation = sales_data[
    [
        "Quantity",
        "Discount",
        "Sales",
        "Profit"
    ]
].corr()

plt.imshow(correlation)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")

plt.show()
```

```python
#Step 11: Detect Outliers
sales_data["Sales"].plot(kind="box")

plt.title("Sales Outlier Detection")
plt.show()
```

```python
#Step 13: Inventory Analysis
#Merge inventory

inventory_data = inventory.merge(
    products,
    on="ProductID"
)
```

```python
#Inventory value:

inventory_data["InventoryValue"] = (
    inventory_data["StockAvailable"]
    *
    inventory_data["CostPrice"]
)
```

```python
#Top warehouses:

warehouse_inventory = (
    inventory_data
    .groupby("Warehouse")["InventoryValue"]
    .sum()
)

warehouse_inventory
```

```python
warehouse_inventory.plot(kind="bar")

plt.title("Inventory Value by Warehouse")
plt.xlabel("Warehouse")
plt.ylabel("Inventory Value")
plt.show()
```

# Business Insights

1. Electronics contributes the highest revenue.

2. A small group of customers contributes a large share of sales.

3. Some products generate high revenue but relatively low profit.

4. Revenue varies across different states.

5. Inventory is concentrated in specific warehouses.

6. A few large orders create noticeable sales outliers.

7. Monthly revenue shows fluctuations, indicating possible seasonal demand.
